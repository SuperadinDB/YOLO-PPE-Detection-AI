```python
from pathlib import Path
from ultralytics import YOLO

PROJECT_ROOT = Path.cwd().parent
SH17_DIR = PROJECT_ROOT / "data" / "SH17 Dataset for PPE Detection"
DATA_YAML = SH17_DIR / "data.yaml"

model = YOLO("yolo11s.pt")

results = model.train(
    data=DATA_YAML,
    epochs=50,
    patience=10,
    imgsz=1024,
    batch=4,
    device=0,
    optimizer="AdamW",
    lr0=0.001,
    cache=False,
    workers=4,

    classes=[
        0,   # person
        2,   # ear-mufs
        8,   # glasses
        9,   # gloves
        10,  # helmet
        14   # shoes
    ],

    project=PROJECT_ROOT / "runs",
    name="sh17_model",
)
```
